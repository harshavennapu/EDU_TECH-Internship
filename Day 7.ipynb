```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
print("Step 2: Loading Datasets...\n")
train_df = pd.read_csv("train_data.csv")
test_df = pd.read_csv("test_data.csv")

print("Train Data Shape:", train_df.shape)
print("Test Data Shape:", test_df.shape)
```

```python
print("Step 3: Preview Train Data\n")
train_df.head()
```

```python
print("Step 4: Dataset Information\n")

print("Train Info:\n")
train_df.info()

print("\nMissing Values:\n")
print(train_df.isnull().sum())

print("\nDuplicate Rows:", train_df.duplicated().sum())
```

```python
print("Step 5: Cleaning Data...\n")

train_df.drop_duplicates(inplace=True)

# Correct syntax
train_df = train_df.ffill()

print("Data cleaned successfully ✅")
```

```python
print("Step 6: Univariate Analysis - Numerical Features\n")

train_df.hist(bins=30, figsize=(12,8))
plt.suptitle("Distribution of Numerical Features")
plt.show()
```

```python
print("Step 7: Category Distribution\n")

sns.countplot(x='Item_Category', data=train_df)
plt.title("Count of Orders by Category")
plt.xticks(rotation=45)
plt.show()
```

```python
print("Columns in dataset:\n")
print(train_df.columns)
```

```python
print("Step 8 : Selling Price Distribution\n")

sns.histplot(train_df['Selling_Price'], bins=30)
plt.title("Selling Price Distribution")
plt.show()
```

```python
print("Step 9: Rating vs Selling Price\n")

sns.scatterplot(x='Item_Rating', y='Selling_Price', data=train_df)
plt.title("Item Rating vs Selling Price")
plt.show()
```

```python
print("Step 10: Price by Category\n")

sns.barplot(x='Item_Category', y='Selling_Price', data=train_df)
plt.title("Average Selling Price by Category")
plt.xticks(rotation=45)
plt.show()
```

```python
print("Step 11: Correlation Heatmap\n")

corr = train_df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
print("Step 12: Outliers in Selling Price\n")

sns.boxplot(x=train_df['Selling_Price'])
plt.title("Boxplot of Selling Price")
plt.show()
```

```python
print("Step 13: Multivariate Analysis - Pairplot\n")

sns.pairplot(train_df)
plt.show()
```

```python
print("Step 14: Test Data Overview\n")

print("Test Info:\n")
test_df.info()

print("\nMissing Values in Test:\n")
print(test_df.isnull().sum())
```
